```python
import pandas as pd
from datetime import datetime, timedelta

```

```python
df = pd.read_excel("DataMaster.xlsx", sheet_name="GAZ_RDN")
```

```python
date = df["trading_date"].min() + timedelta(days=1)
date = date.date()
```

```python
date_list_rdn = []
date_list_otf = []
while date < datetime.today().date():
    if date.weekday() < 5:  # Check if the date is a weekday (0-4)
        date_list_otf.append(date)
    date_list_rdn.append(date)
    date += timedelta(days=1)

```
